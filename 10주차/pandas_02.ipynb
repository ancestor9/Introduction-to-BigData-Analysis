### **The anatomy of a boxplot**
<img src='https://api.www.labxchange.org/api/v1/xblocks/lb:LabXchange:d8863c77:html:1/storage/211626365402575-b88c4d0fdacd5abb4c3dc2de3bc004bb.png' height=350>

- Boxplots are a good way to visualize data sets since they display a lot of important information in a simple, standard format.
- Boxplots describe data sets using 5 particular numbers: the minimum, first quartile, median, third quartile, and maximum.
- The box in the diagram begins with the first quartile and ends with the third quartile, with the median indicated as a line that divides the box.
- Lines extend from the first quartile down to the minimum and from the third quartile up to the maximum. 25% of the data values in the data set fall between the minimum and the first quartile; 25% of the data values fall between the first quartile and the median; 25% of the data values fall between the median and the third quartile; 25% of the data values fall between the third quartile and the maximum.

## 📊 히스토그램 특징
- **연속형 데이터의 분포를 시각적으로 표현하는 데 사용되는 그래프입니다. 막대 그래프로 그려지며, 각 막대는 특정 구간(bins)에 해당하는 데이터의 개수(도수, frequency) 시각화**
### ✅ 연속형 변수에 적합
- **예:** 키, 몸무게, 시험 점수, 온도 등
- **주의:** 범주형 변수에는 적합하지 않음 → *막대그래프 사용*

---

### ✅ 데이터의 분포 확인 가능
- 정규분포 여부
- 왜도(skewness) 파악
- 이상치(outlier) 탐지
- 데이터가 집중된 구간 시각화

---

### ✅ Bin 수에 따라 해석 달라짐
- Bin 수가 **너무 적으면** → 분포가 **거칠게** 보임
- Bin 수가 **너무 많으면** → **노이즈**가 많아져 해석이 어려움


```python
import matplotlib.pyplot as plt
import seaborn as sns

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

sns.boxplot(data=air_quality, y='station_paris', ax=axes[0])
axes[0].set_title('Boxplot of Station Paris')

sns.histplot(data=air_quality, x='station_paris', ax=axes[1])
axes[1].set_title('Histogram of Station Paris')

plt.tight_layout()
plt.show()
```

```python
# Select only numeric columns for the histogram
numeric_cols = air_quality.select_dtypes(include=['float64', 'int64']).columns

# Create subplots for each numeric column in a single row
fig, axes = plt.subplots(1, len(numeric_cols), figsize=(len(numeric_cols) * 5, 5))

# Plot histograms for each numeric column
for i, col in enumerate(numeric_cols):
    sns.histplot(data=air_quality, x=col, ax=axes[i])
    axes[i].set_title(f'Histogram of {col}')

plt.tight_layout()
plt.show()
```

```python
# prompt: url을 데이터프레임으로 읽기

import pandas as pd
# The URL points to the HTML page on GitHub, not the raw CSV data.
# Need to modify the URL to get the raw data link.
url_raw = 'https://raw.githubusercontent.com/pandas-dev/pandas/main/doc/data/air_quality_long.csv'

air_quality = pd.read_csv(url_raw, index_col="date.utc", parse_dates=True)
air_quality.head()
```

```python
# prompt: air_quality의 각 컬럼의 고유값, 결측치, 결측비율, 대표값 5개를 데이터프레임으로 생성하는 함수를 간단하게 만드랄

import pandas as pd
def summarize_dataframe(df):
    """
    Generates a summary DataFrame containing information about each column in the input DataFrame.

    For each column, the summary includes:
    - 'Unique Values': The number of unique values.
    - 'Missing Values': The number of missing values (NaN).
    - 'Missing Percentage': The percentage of missing values.
    - 'Representative Values': The first 5 unique values.

    Args:
        df (pd.DataFrame): The input DataFrame to summarize.

    Returns:
        pd.DataFrame: A DataFrame with columns as index and summary statistics as columns.
    """
    summary_data = []
    for col in df.columns:
        unique_values = df[col].nunique()
        missing_values = df[col].isnull().sum()
        missing_percentage = (missing_values / len(df)) * 100
        representative_values = df[col].dropna().unique().tolist()[:5]

        summary_data.append({
            'Unique Values': unique_values,
            'Missing Values': missing_values,
            'Missing Percentage': missing_percentage,
            'Representative Values': representative_values
        })

    summary_df = pd.DataFrame(summary_data, index=df.columns)
    return summary_df

# Example usage with the air_quality DataFrame:
air_quality_summary = summarize_dataframe(air_quality)
air_quality_summary
```
