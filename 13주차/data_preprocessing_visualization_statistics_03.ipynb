### 공원, 역, 사거리, 대학교 근처 장소를 중심으로 분석

```python
# Rental_Station 컬럼에서 공원, 역, 사거리, 대학교 등의 text가 포함되었는지 여부를 특성 컬럼으로 만들어줘

df['Is_Park'] = df['Rental_Station'].str.contains('공원')
df['Is_Station'] = df['Rental_Station'].str.contains('역')
df['Is_Intersection'] = df['Rental_Station'].str.contains('사거리')
df['Is_University'] = df['Rental_Station'].str.contains('대학교')

display(df.head())
```

```python
# 추가된  특성컬럼별로 Rental_Count의 합계를 구해줘

rental_count_by_location_type = df[['Is_Park', 'Is_Station', 'Is_Intersection', 'Is_University', 'Rental_Count']].groupby(['Is_Park', 'Is_Station', 'Is_Intersection', 'Is_University']).sum()

display(rental_count_by_location_type)
```

## **정규표현식(regular expression)**

```python
# 정규표현식(regular expression)을 학습하려고해 예제 코드를 제시해줘

import re

# Example string
text = "The quick brown fox jumps over the lazy dog. 12345"
```

```python
# 1. Basic matching
# Find all occurrences of "the" (case-insensitive)
matches_the = re.findall(r"the", text, re.IGNORECASE)
print(f"Occurrences of 'the': {matches_the}")
```

```python

# 2. Matching digits
# Find all digits
matches_digits = re.findall(r"\d", text)
print(f"Digits: {matches_digits}")
```

```python

# 3. Matching words
# Find all words (sequences of letters)
matches_words = re.findall(r"\b[a-zA-Z]+\b", text)
print(f"Words: {matches_words}")
```

```python

# 4. Matching specific patterns
# Find a sequence of 5 digits
match_five_digits = re.search(r"\d{5}", text)
if match_five_digits:
    print(f"Five consecutive digits found: {match_five_digits.group()}")
```

```python

# 5. Replacing patterns
# Replace spaces with hyphens
text_replaced = re.sub(r"\s", "-", text)
print(f"Text with spaces replaced: {text_replaced}")
```

```python

# 6. Splitting strings
# Split the string by spaces
split_text = re.split(r"\s", text)
print(f"Split text: {split_text}")
```
